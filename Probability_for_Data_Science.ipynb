```python
#Import Libraries
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python
# ── 1. Basic Probability ───────────────────────────────────────────
print("=" * 45)
print("   BASIC PROBABILITY")
print("=" * 45)

np.random.seed(42)
flips = np.random.choice(['H', 'T'], size=10000)
p_heads = np.sum(flips == 'H') / len(flips)
print(f"Theoretical P(heads)  : 0.5000")
print(f"Experimental P(heads) : {p_heads:.4f}  (10,000 flips)")
print(f"Law of Large Numbers  : experimental → theoretical ✓")
```

```python
# ── 2. Conditional Probability and Bayes' Theorem ─────────────────
print("\n" + "=" * 45)
print("   BAYES THEOREM (Spam Example)")
print("=" * 45)

total_emails = 1000
spam_emails  = 200
free_in_spam = 180
free_total   = 190

p_spam            = spam_emails / total_emails
p_free_given_spam = free_in_spam / spam_emails
p_free            = free_total / total_emails
p_spam_given_free = (p_free_given_spam * p_spam) / p_free

print(f"P(spam)              : {p_spam:.4f}")
print(f"P(FREE | spam)       : {p_free_given_spam:.4f}")
print(f"P(spam | FREE)       : {p_spam_given_free:.4f}")
print(f"→ If email has FREE, there is a {p_spam_given_free*100:.1f}% chance it is spam")
```

```python
# ── 3. Normal Distribution ────────────────────────────────────────
print("\n" + "=" * 45)
print("   NORMAL DISTRIBUTION (68-95-99.7 Rule)")
print("=" * 45)

mu, sigma = 50, 10
dist = stats.norm(mu, sigma)
print(f"Mean (μ) = {mu},  Std Dev (σ) = {sigma}")
print(f"P(40 < X < 60)  : {dist.cdf(60) - dist.cdf(40):.4f}  → within 1σ  [expect 0.6827]")
print(f"P(30 < X < 70)  : {dist.cdf(70) - dist.cdf(30):.4f}  → within 2σ  [expect 0.9545]")
print(f"P(20 < X < 80)  : {dist.cdf(80) - dist.cdf(20):.4f}  → within 3σ  [expect 0.9973]")
```

```python
# ── 4. Binomial Distribution ──────────────────────────────────────
print("\n" + "=" * 45)
print("   BINOMIAL DISTRIBUTION")
print("=" * 45)

n, p = 10, 0.3
binom = stats.binom(n, p)
print(f"n={n} users, p={p} click rate")
print(f"P(exactly 3 click)  : {binom.pmf(3):.4f}")
print(f"P(at least 5 click) : {1 - binom.cdf(4):.4f}")
print(f"Expected clicks     : {binom.mean():.1f}")
```

```python
# ── 5. Poisson Distribution ───────────────────────────────────────
print("\n" + "=" * 45)
print("   POISSON DISTRIBUTION")
print("=" * 45)

lam = 5
poisson = stats.poisson(lam)
print(f"Average rate (λ) = {lam} requests/second")
print(f"P(exactly 5)     : {poisson.pmf(5):.4f}")
print(f"P(exactly 8)     : {poisson.pmf(8):.4f}")
print(f"P(more than 10)  : {1 - poisson.cdf(10):.4f}")
```
